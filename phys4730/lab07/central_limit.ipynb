```python
%reset -f

import numpy as np
import scipy.optimize as SciOpt
from numpy import random
from numpy import linalg as LA
import matplotlib.pyplot as plt
import matplotlib
from scipy import stats

matplotlib.rcParams.update({'font.size': 18})
```

```python
### generate a uniform distribution ###

aa = 0.5
W  = 1.0
nbins = 100
N = 1000000
x = random.uniform(aa,aa+W,size=N)
xmean = np.mean(x)
xsigm = np.std(x)
```

```python
### make a plot ###

Deltax = W/nbins

plt.figure(figsize=(10,6))

n,b,p = plt.hist(x, bins=nbins, range=[aa,aa+W], density=True)
plt.xlabel("$x$")
plt.ylabel("pdf")
plt.grid()

dist = stats.uniform(aa,W)    # arguments are left edge, width
xx   = np.arange(aa,aa+W,Deltax)
pdf  = dist.pdf(xx)
pdfmax = np.max(pdf)

plt.plot(xx,pdf,color='r')

textm = "$\overline{x}$  = %9.5f" % xmean
texts = "$\sigma_{x}$ = %9.5f"   % xsigm

plt.text(xmean+0.05*W,0.52*pdfmax,textm)
plt.text(xmean+0.05*W,0.46*pdfmax,texts)

xmean_exp = aa + 0.5*W
xsigm_exp = W/np.sqrt(12.0)

plt.show()

print("expected mean    = %9.5f" % xmean_exp)
print("expected std dev = %9.5f" % xsigm_exp)
```

```python
### now, check central limit theorem ###

aa = 0.5
W  = 1.0
nbins = 100
N  = 12

Ntrials = 100000
xmeanN = np.zeros(Ntrials)

### loop over trials, each time sampling N values of x ###

for i in np.arange(0,Ntrials):
    
    x = random.uniform(aa,aa+W,size=N)
    xmeanN[i] = np.mean(x)
```

```python
### plot the means ###

plt.figure(figsize=(10,6))

n,b,p = plt.hist(xmeanN, bins=nbins, range=[aa,aa+W], density=True)
plt.xlabel("$x$")
plt.ylabel("pdf")
plt.grid()

mu    = xmean_exp
sigma = xsigm_exp/np.sqrt(N)

dist = stats.norm(mu,sigma)
xx   = np.arange(aa,aa+W,Deltax/10.0)
pdf  = dist.pdf(xx)
pdfmax = np.max(pdf)

plt.plot(xx,pdf,color='r')
textN = "N = %3d" % N
plt.title(textN)

plt.show()

print("mean    = %9.5f" % mu)
print("std dev = %9.5f" % sigma)
```
