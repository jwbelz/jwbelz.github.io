```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar
from scipy.stats import chi2, kstest, ks_2samp, norm
```

```python
### read data with pandas. display number of sources and header ###

CACHE = 'hyades_gaia_dr3.csv'
df = pd.read_csv(CACHE)
print(f'{len(df):,} quality-screened sources')
df.head()
```

```python
### basic plotting of database quantities ### 

# plot the parallax in milli-arcseconds

plt.figure(figsize=(6,6))

plt.hist(df.parallax,bins=50,range=(0,50), density=True, histtype='step')
plt.xlabel("parallax (mas)")
plt.ylabel("density")
plt.grid()
plt.show()
```

```python
### example use of pandas to subdivide data ###

# suppose we're interested in 1 mas < parallax < 5 mas

interesting = df.parallax.between(1.0,5.0)
cutdata     = df.loc[interesting].copy()

# plot the cumulative distribution for proper motion in right ascension (pmra)

plt.figure(figsize=(6,6))

plt.hist(cutdata.pmra,bins=50,range=(-50,50), density=True, cumulative=True, histtype='step')
plt.xlabel(r"$\mu_{\alpha *}$")
plt.ylabel("ECDF")
plt.grid()
plt.show()
```
