```python
# Now, let's do some linear fitting of points
# using Vandermonde inversion. 

from numpy.linalg import pinv
from numpy.linalg import inv

# Make up some data...

# ...independent variable
xaxis = np.arange(1.0,11.0,1.0)

# ...dependent variable
yaxis = 2.0 + 0.5*xaxis + np.random.normal(0.0, 0.2, 10)

plt.scatter(xaxis,yaxis)
plt.xlabel("Independent variable $x$")
plt.ylabel("Dependent variable $y$")
plt.show()
```

```python
# Set up the Vandermonde matrix

# we need the x and y-axes as column vectors
xcv = xaxis[:, None]
ycv = yaxis[:, None]

# now can create the matrix
vdm = np.concatenate((xcv**0, xcv**1), axis=1)

print(vdm)
```

```python
# solve to obtain best fit
a = np.transpose(pinv(vdm).dot(ycv))[0]
print(a)
```

```python
# make a plot
plt.scatter(xaxis,yaxis)
xx = np.arange(xaxis[0],xaxis[9],0.01)
yy = a[0] + a[1]*xx
plt.plot(xx,yy)
plt.xlabel("Independent variable $x$")
plt.ylabel("Dependent variable $y$")
plt.show()
```

```python
# now, suppose two independent variables 

from numpy.linalg import pinv

x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)

# np.meshgrid() basically creates all possible pairings
# of x and y values which makes calculating z values easy

xv, yv = np.meshgrid(x, y)

zv = 1.0 + 2.0*xv + 3.0*yv

plt.imshow(zv)
plt.colorbar()
plt.xlabel("Independent variable $x$")
plt.ylabel("Independent variable $y$")
plt.jet()
```

```python
# add some random noise 

zv = zv + np.random.normal(0, 1, (100,100))

plt.imshow(zv)
plt.colorbar()
plt.xlabel("Independent variable $x$")
plt.ylabel("Independent variable $y$")
plt.jet()
```

```python
# set up the Vandermonde matrix

# we need the xv, yv and zv as column vectors
xv=np.reshape(xv, -1); xv=xv[:, None]
yv=np.reshape(yv, -1); yv=yv[:, None]
zv=np.reshape(zv, -1); zv=zv[:, None]

# create the matrix
vdm=np.concatenate((xv**0, xv, yv), axis=1)
print(vdm)
```

```python
# now we can solve
a = np.transpose(pinv(vdm).dot(zv))[0]
print(a)
```
