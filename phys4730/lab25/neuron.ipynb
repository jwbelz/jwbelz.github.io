```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams.update({'font.size': 14})
matplotlib.use('nbAgg')
```

We want to train the neuron on a straight line, $y(x) = 2x + 1$.

```python
# Training data: y = 2x + 1

X = np.array([[0], [1], [2], [3]], dtype=float)  # inputs
y = np.array([[1], [3], [5], [7]], dtype=float)  # outputs
```

```python
# Initialize weights and bias randomly

W = np.random.uniform(low=1.0, high=3.0, size=1)[0]
b = np.random.uniform(low=0.0, high=2.0, size=1)[0]
```

```python
# Learning rate

lr = 0.01
```

```python
# storage arrays - JB

nepochs = 1000

Wstore  = np.zeros(nepochs)
bstore  = np.zeros(nepochs)
estore  = np.zeros(nepochs)
lstore  = np.zeros(nepochs)
```

```python
# Training loop

for epoch in np.arange(nepochs):

    # Forward pass
    y_pred = X * W + b

    # Compute loss (Mean Squared Error)
    loss = np.mean((y_pred - y) ** 2)

    # Compute gradients (derivatives)
    dW = np.mean(2 * X * (y_pred - y))
    db = np.mean(2 * (y_pred - y))

    # Update weights
    W -= lr * dW
    b -= lr * db

    Wstore[epoch] = W
    bstore[epoch] = b
    estore[epoch] = epoch
    lstore[epoch] = loss

    # Print progress occasionally
    if epoch % 200 == 0:
        print("Epoch: %4d Loss: %10.8f W: %10.8f b: %10.8f" % (epoch,loss,W,b) )
```

```python
# Test the trained model

x_test = np.arange(-5.0,6.0,1.0)
y_test = x_test * W + b

for i in np.arange(0,len(x_test)):
    print(" input x: %7.3f     predicted y: %7.3f" % (x_test[i],y_test[i]) )
```

```python
# plot learning curve

plt.figure(figsize=(8, 8))

plt.subplot(2,1,1)

plt.plot(estore,lstore)
plt.ylabel("loss function")
plt.xlim([-100.0,1100.0])
plt.yscale("log")
plt.title("Neuron Training Progress")
plt.grid()

plt.subplot(2,1,2)

plt.plot(estore,Wstore, label="W")
plt.plot(estore,bstore, label="b")
plt.xlabel("epoch")
plt.ylabel("parameter")
plt.axis([-100.0,1100.0,0.0,3.0])
plt.grid()
plt.legend()


plt.tight_layout()
plt.show()
```
