```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('nbAgg')
```

We want to train the network to output a sine wave, $y=\sin{x}$. 

```python
def ytrain(xx):
    return np.sin(xx)    
```

```python
# Training data

N = 200
x = np.linspace(-2*np.pi, 2*np.pi, N).reshape(-1,1) # N-element column vector
y = ytrain(x)                                       # N-element columm vector 
```

two-hidden, one output NN architecture

Weights:

    * W1: (2x1) hidden weights
    * b1: (2,)  hidden biases
    * W2: (1x2) output weights
    * b2: (1,)  output biases

```python
W1 = np.random.randn(2,1) * 0.5   # shape: (2,1) 
b1 = np.zeros((2,1))              # shape: (2,1)
W2 = np.random.randn(1,2) * 0.5   # shape: (1,2)
b2 = np.zeros((1,1))              # shape: (1,1)

if(True):
    print(W1)
    print(b1)
    print(W2)
    print(b2)
```

```python
# activation function and its derivative

def tanh(z):
    return np.tanh(z)

def dtanh(z):
    return 1 - np.tanh(z)**2
```

```python
# parameters and arrays for training

lr     =  0.01
epochs =  5000

lr     =  0.10
epochs = 10000

estore   = np.zeros(epochs)
lstore   = np.zeros(epochs)
```

```python
# Training loop

for epoch in range(epochs):

    # Forward pass
    
    z1 = W1 @ x.T + b1       # shape: (2,N)
    h  = tanh(z1)            # shape: (2,N)
    y_pred = W2 @ h + b2     # shape: (1,N)

    # Compute and store loss
    
    loss = np.mean((y_pred - y.T)**2)

    estore[epoch] = epoch
    lstore[epoch] = loss
    
    # Backpropagation
    
    dL_dy = 2*(y_pred - y.T)/N

    dW2 = dL_dy @ h.T                          # (1,2)
    db2 = np.sum(dL_dy, axis=1, keepdims=True) # (1,1)

    dh = W2.T @ dL_dy                          # (2,N)
    dz1 = dh * dtanh(z1)                       # (2,N)

    dW1 = dz1 @ x                              # (2,N)
    db1 = np.sum(dz1, axis=1, keepdims=True)   # (2,N)
 
    # Update
    
    W1 -= lr * dW1   # shape: (2,1) 
    b1 -= lr * db1   # shape: (2,1) 
    W2 -= lr * dW2   # shape: (1,2) 
    b2 -= lr * db2   # shape: (1,1) 

    # debuggin'
    
    if(False):
        print("z1:",z1)
        print("h:",h)
        print("y_pred",y_pred)
        print("dL_dy:",dL_dy)
        print("dW2:",dW2)
        print("db2:",db2)
        print("dh:",dh)
        print("dz1:",dz1)
        print("dW1:",dh)
        print("db1:",dz1)
        print(W1)
        print(b1)
        print(W2)
        print(b2)

    if epoch % 1000 == 0:
        print(f"epoch {epoch:5d}, loss={loss:.5f}")
```

```python
# plot results

z1 = W1 @ x.T + b1
h = tanh(z1)
y_pred = (W2 @ h + b2).T

plt.figure(figsize=(8,8))

plt.subplot(2,1,1)

plt.plot(x, y, 'k', label="sin(x)")
plt.scatter(x, y_pred, c='r', s=10, label="NN output")
plt.legend(loc="upper right")

plt.subplot(2,1,2)

plt.plot(x,y_pred-y, 'k', label="residual")
plt.legend(loc="upper right")
plt.xlabel("$x$")
plt.ylabel("$y_{pred} - \sin{x}$")
plt.tight_layout()
plt.show()
```

```python
# plot learning curve

plt.figure(figsize=(8, 5))

plt.subplot(1,1,1)

plt.plot(estore,lstore)
plt.title("Network Training Progress")
plt.xlabel("epoch")
plt.ylabel("loss function")
plt.yscale("log")
#plt.axis([0,epochs,0.0,1.0])
plt.grid()
plt.tight_layout()
plt.show()
```
