```python
%reset -f

import numpy as np
from scipy.optimize import curve_fit
from scipy.special  import gammaincc
import matplotlib.pyplot as plt
```

```python
### the model function ###                             
        
def func(t,N0,tau):

    return N0*np.exp(-t/tau)
```

```python
### Model parameters ###

T     = 500
dt    = 10
N0    = 1000
tau   = 100
stdev = 20
```

```python
### Create the artificial dataset ###

nobs   = int(T/dt + 1.5)                                      
t      = dt*np.arange(nobs)
N      = func(t,N0,tau)
Nfluct = stdev*np.random.normal(size=nobs)
N      = N + Nfluct
sig    = np.zeros(nobs) + stdev
```

```python
### Plot the data ###

plt.errorbar(t, N, yerr=sig, fmt = 'o', label='"data"')
plt.xlabel("time (some units)")
plt.ylabel("counts per time bin")
plt.legend()
plt.show()
```
