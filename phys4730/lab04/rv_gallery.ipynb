A tour of different random variable distributions. 

```python
%reset -f

import numpy as np
from numpy import random
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams.update({'font.size': 16})
```

```python
### uniform distribution ###

aa = 23.0
bb = 25.0
nbins = 100
N = 1000000
x = random.uniform(aa,bb,size=N)

# plots

plt.figure(figsize=(7,7))
           
n,b,p = plt.hist(x, bins=nbins, range=[aa,bb])
plt.xlabel("$x$")
plt.ylabel("counts/bin")
plt.grid()
plt.show()
```

```python
### Now, weigh events by 1/N ###

plt.figure(figsize=(7,7))

n,b,p = plt.hist(x, bins=nbins, range=[aa,bb], weights=0*x+1/N)
plt.xlabel("$x$")
plt.ylabel("probability/bin")
plt.grid()

plt.show()
```

```python
### Now, weigh events by 1/(N*Delta-x) ###
### This gives a data-derived estimate ###
### for the pdf, or an "empirical pdf" ###

Deltax = (bb-aa)/nbins

plt.figure(figsize=(7,7))

n,b,p = plt.hist(x, bins=nbins, range=[aa,bb], weights=0*x + 1.0/(N*Deltax))
plt.xlabel("$x$")
plt.ylabel("pdf")
plt.grid()

from scipy import stats
dist = stats.uniform(aa,(bb-aa))    # arguments are left edge, width

# compare the 'true' pdf
    
xx = np.arange(aa,bb,Deltax)
pdf  = dist.pdf(xx)
plt.plot(xx,pdf,color='r')

plt.show()
```

```python
### Another way of doing the same thing ###

Deltax = (bb-aa)/nbins

plt.figure(figsize=(7,7))

n,b,p = plt.hist(x, bins=nbins, range=[aa,bb], density=True)  # <--- this call changed 
plt.xlabel("$x$")
plt.ylabel("pdf")
plt.grid()

from scipy import stats
dist = stats.uniform(aa,(bb-aa))    # arguments are left edge, width
 
xx = np.arange(aa,bb,Deltax)
pdf = dist.pdf(xx)
plt.plot(xx,pdf,color='r')

plt.show()
```

```python
# Gaussian distribution 

aa = -10.0
bb =  10.0
nbins = 200
Deltax = (bb-aa)/nbins
N = 1000000

mu, sigma = 1.0, 2.0   # mean and standard deviation
x = random.normal(mu, sigma, size=N)

# plot

plt.figure(figsize=(7,7))

n,b,p = plt.hist(x, bins=nbins, range=[aa,bb],density=True)  
plt.xlabel('$x$')
plt.ylabel('pdf')
plt.grid()

from scipy import stats
dist = stats.norm(mu,sigma)

xx = np.arange(-10.0,10.0,Deltax)
pdf = dist.pdf(xx)              # p(x) = 1/sqrt(2*pi*sigma**2)*exp(-0.5*(x-mu)/sigma**2)
plt.plot(xx,pdf,color='r')

plt.show()
```

```python
# Binomial distribution

n = 100
p = 0.4
x = random.binomial(n, p, size=N)

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=101, range=[-0.5,100.5]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.grid()

plt.show()
```

```python
# plot again, zooming in on x axis...
# (making the point that binomial distribution is discrete)

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=101, range=[25.5,30.5]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.grid()

plt.show()
```

```python
# Bernoulli distribution (Binomial with n = 1)

# In probability theory and statistics, the Bernoulli distribution, 
# named after Swiss mathematician Jacob Bernoulli, is the discrete 
# probability distribution of a random variable which takes the 
# value 1 with probability p and the value 0 with probability q=1-p.

# In Python, one asks for binomial-distributed data points
# with n=1, corresponding to a single coin flip at a time

n = 1
p = 0.4
x = random.binomial(n, p, size=N)

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=11, range=[-0.05,1.05]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.grid()

plt.show()
```

```python
# Poisson distribution
# (also discrete, only integers allowed)

mu = 2.3
x  = random.poisson(mu, size=N)

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=11, range=[-0.5,10.5]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.grid()

plt.show()
```

```python
# log-normal distribution

mu  = 1.0
sig = 0.3

x  = random.lognormal(mu,sig, size=N)

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=100, range=[0.0,10.0]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.grid()

plt.show()
```

```python
# Weibull distribution

gmm = 1.5

x  = random.weibull(gmm, size=N)

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=100, range=[0.0,10.0]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.grid()

plt.show()
```

```python
# exponential distribution

beta = 2.0

x  = random.exponential(beta, size=N)

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=100, range=[0.0,10.0]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.grid()
```

```python
# Cauchy/Lorentzian distribution

x  = random.standard_cauchy(size=N)

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=200, range=[-1000.0,1000.0]) 
plt.xlabel('$x$')
plt.ylabel('counts/bin')
plt.yscale('log')
plt.grid()

plt.show()
```

```python
# Pareto distribution

# np.random.pareto() draws random samples from a Pareto II 
# or Lomax distribution with a specified shape. The Pareto II 
# distribution is a shifted Pareto distribution. By adding 1 
# and multiplying by the scale parameter x_m, the classical 
# Pareto distribution can be obtained from the Lomax distribution.

x_m, alpha = 1, 1.16           # alpha = 1.16 corresponds to 80-20 rule

x  = (random.pareto(alpha,size=N) + 1.0)*x_m

# plot

plt.figure(figsize=(5,5))
plt.subplot(1,1,1)

n,b,p = plt.hist(x, bins=200, range=[0.0,200.0], density=True) 
plt.xlabel('$x$')
plt.ylabel('p(x)')
plt.yscale('log')
plt.grid()

plt.show()
```
