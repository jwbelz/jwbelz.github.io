```python
import numpy as np

import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

font = {'size'   : 14}
matplotlib.rc('font', **font)
matplotlib.rc('xtick', labelsize=14) 
matplotlib.rc('ytick', labelsize=14) 
matplotlib.rcParams['figure.dpi'] = 100

import pandas as pd 

import sklearn
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
from sklearn import neighbors
```

```python
### read data from file, drop redundant first column ###

LearningSet = pd.read_csv('HPLearningSet.csv')
LearningSet = LearningSet.drop(LearningSet.columns[0], axis=1) #We want to drop the first column of the file
LearningSet
```

```python
### pick train and test sets as in DT example ###

TrainSet = LearningSet.iloc[:13,:]
TestSet  = LearningSet.iloc[13:,:]
```

```python
### split in features and labels ###

Xtrain = TrainSet.drop(['P_NAME','P_HABITABLE'],axis=1)
Xtest  = TestSet.drop(['P_NAME','P_HABITABLE'],axis=1)  

ytrain = TrainSet.P_HABITABLE
ytest  = TestSet.P_HABITABLE
```

```python
### deploy kNN ###

model = neighbors.KNeighborsClassifier(n_neighbors = 3)
model.fit(Xtrain.iloc[:,:2],ytrain) # just fit two features, for starters
```

```python
### use the fitted model to predict the labels for the test set ###
### compare the results                                         ###

ytestpred = model.predict(Xtest.iloc[:,:2])
    
print("prediction:  %4d %4d %4d %4d %4d" % \
      (ytestpred[0], ytestpred[1],ytestpred[2],ytestpred[3],ytestpred[4])) 

print("observation: %4d %4d %4d %4d %4d" % \
      (ytest.values[0], ytest.values[1],ytest.values[2],ytest.values[3],ytest.values[4])) 
     
```

```python
### check accuracy score ###

print(metrics.accuracy_score(ytrain, model.predict(Xtrain.iloc[:,:2])))
print(metrics.accuracy_score(ytest,  model.predict(Xtest.iloc[:,:2])))    
```

```python
### we can access the found nearest neighbors as follows ###

mk = model.kneighbors(Xtest.iloc[:,:2])  
mk       
```

#### Well and good, but we haven't scaled the learning set to properly perform kNN!


```python
### important: we only scale the training set ###

scaler = sklearn.preprocessing.RobustScaler()
scaler.fit(Xtrain)                         
scaledXTrain = scaler.transform(Xtrain)
scaledXTrain
```

```python
### compare the unscaled version ###

Xtrain
```

```python
scaledXtest = scaler.transform(Xtest) # note that these are now numpy arrays, not data frames
```

```python
### deploy kNN on the scaled features ###

model.fit(scaledXTrain[:,:2],ytrain)
```

```python
### use the fitted model to predict the labels for the test set ###
### compare the results                                         ###

ytestpred = model.predict(scaledXtest[:,:2])
    
print("prediction:  %4d %4d %4d %4d %4d" % \
      (ytestpred[0], ytestpred[1],ytestpred[2],ytestpred[3],ytestpred[4])) 

print("observation: %4d %4d %4d %4d %4d" % \
      (ytest.values[0], ytest.values[1],ytest.values[2],ytest.values[3],ytest.values[4])) 
     
```

```python
### check accuracy score ###

print(metrics.accuracy_score(ytrain, model.predict(scaledXTrain[:,:2])))
print(metrics.accuracy_score(ytest,  model.predict(scaledXtest[:,:2])))    
```
