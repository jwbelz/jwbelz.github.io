```python
%reset -f 

import numpy as np
import scipy.optimize as SciOpt
import scipy.stats as SciStat
import scipy.special as SciSpec
from numpy import random as rnd
from numpy import linalg as LA
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rcParams.update({'font.size': 14})
```

```python
### define the domain ###

xmin = -10.0
xmax =  10.0
dx   =   0.001

x = np.arange(xmin,xmax,dx)
```

```python
### specify parameters for two RVs ###

if(True):

    # Case 1

    thet1, sigm1 = 1.0, 0.3
    thet2, sigm2 = 2.0, 0.3

else:
    
    # Case 2

    thet1, sigm1 = 2.0, 0.3
    thet2, sigm2 = 2.0, 0.6
```

```python
### create various functions for plotting ###

### two pdfs ###

pdf1 = np.exp(-0.5*(x-thet1)**2/sigm1**2)/(np.sqrt(2.0*np.pi*sigm1*sigm1))
pdf2 = np.exp(-0.5*(x-thet2)**2/sigm2**2)/(np.sqrt(2.0*np.pi*sigm2*sigm2))

### log of the pdfs ###

lnf1 = -np.log(np.sqrt(2.0*np.pi)*sigm1) - 0.5*(x-thet1)**2/sigm1**2
lnf2 = -np.log(np.sqrt(2.0*np.pi)*sigm2) - 0.5*(x-thet2)**2/sigm2**2

### derivative of ln(f) wrt theta ###

lnfp1 = (x-thet1)/sigm1**2
lnfp2 = (x-thet2)/sigm2**2

### square the term above ###

lnfp21 = lnfp1*lnfp1
lnfp22 = lnfp2*lnfp2

### integrand of fisher information ###

igrand1 = lnfp21 * pdf1
igrand2 = lnfp22 * pdf2

### numerically compute the integrals ###

fishi1 = np.sum(igrand1*dx)
fishi2 = np.sum(igrand2*dx)
```

```python
### Make the figure ###

figsizex =  8.5   # units are inches
figsizey = 11.0   # units are inches

fig = plt.figure(figsize=(figsizex,figsizey))

# panel 1

fig.add_subplot(511)

plt.plot(x,pdf1,color='k',label="pdf 1")
plt.plot(x,pdf2,color='r',label="pdf 2")
plt.ylabel("PDF")
plt.legend(loc="upper left")

# panel 2

fig.add_subplot(512)

plt.plot(x,lnf1,color='k')
plt.plot(x,lnf2,color='r')
plt.ylabel("$\ln{(PDF)}$")

# panel 3

fig.add_subplot(513)

plt.plot(x,lnfp1,color='k')
plt.plot(x,lnfp2,color='r')
plt.ylabel("$\ln{(PDF)}'$")

# panel 4

fig.add_subplot(514)

plt.plot(x,lnfp21,color='k')
plt.plot(x,lnfp22,color='r')
plt.ylabel("$(\ln{(PDF)')^2}$")

# panel 5

fig.add_subplot(515)

plt.plot(x,igrand1,color='k')
plt.plot(x,igrand2,color='r')
plt.xlabel("$x$")
plt.ylabel("integrand")

txtstr1 = "$I_1 = %8.5f $" % fishi1
txtstr2 = "$I_2 = %8.5f $" % fishi2

ymax1 = np.max(igrand1)
ymax2 = np.max(igrand2)
ymax  = np.max([ymax1,ymax2])

plt.text(-9.0,0.6*ymax,txtstr1,c='k')
plt.text(-9.0,0.3*ymax,txtstr2,c='r')

plt.savefig("fisher.png",dpi=300,facecolor='white', transparent=False)
plt.show()
```
