Given $x$, the neuron computes $\hat{y} = Wx + b$

The training target is the best fit $y(x)$ of a straight-line fit to the data.

We define the <b>loss function</b> ${\cal L}$ as the mean-squared error
$$
{\cal L} = \overline{(\hat{y} - y)^2}
$$

Then we can use <b>back propagation</b> to improve our guesses for $W$, $b$ as follows:

$$
\frac{\partial {\cal L}}{\partial W} = \frac{\partial {\cal L}}{\partial \hat{y}} \frac{\partial \hat{y}}{\partial W} = \overline{2 (\hat{y} - y) x}
$$

$$
\frac{\partial {\cal L}}{\partial b} = \frac{\partial {\cal L}}{\partial \hat{y}} \frac{\partial \hat{y}}{\partial b} = \overline{2 (\hat{y} - y)}
$$

$$
W_{new} = W - \eta \frac{\partial {\cal L}}{\partial W} = W - \overline{2 \eta (\hat{y} - y) x}
$$

$$
b_{new} = b - \eta \frac{\partial {\cal L}}{\partial b} = b - \overline{2 \eta (\hat{y} - y)}
$$

where $0 < \eta < 1$ is known as the <b>learning rate</b>.

Now use the neuron to perform a linear fit

```python
# Initialize weights and bias randomly

W = np.random.uniform(low=1.0, high=3.0, size=1)[0]
b = np.random.uniform(low=0.0, high=2.0, size=1)[0]
```

```python
# Learning rate

lr = 0.01
```

```python
# storage arrays

nepochs = 1000

Wstore  = np.zeros(nepochs)
bstore  = np.zeros(nepochs)
estore  = np.zeros(nepochs)
lstore  = np.zeros(nepochs)
```

```python
# Training loop

for epoch in np.arange(nepochs):

    # Forward pass
    y_pred = X * W + b

    # Compute loss (Mean Squared Error)
    loss = np.mean((y_pred - y) ** 2)

    # Compute gradients (derivatives)
    dW = np.mean(2 * X * (y_pred - y))
    db = np.mean(2 * (y_pred - y))

    # Update weights
    W -= lr * dW
    b -= lr * db

    Wstore[epoch] = W
    bstore[epoch] = b
    estore[epoch] = epoch
    lstore[epoch] = loss

    # Print progress occasionally
    if epoch % 200 == 0:
        print("Epoch: %4d Loss: %10.8f W: %10.8f b: %10.8f" % (epoch,loss,W,b) )
```

```python
# Print the fit results

print("neuron best fit for slope:    ",W)
print("neuron best fit for intercept:",b)
```

```python
# plot learning curve

plt.figure(figsize=(8, 8))

plt.subplot(2,1,1)

plt.plot(estore,lstore)
plt.ylabel("loss function")
plt.xlim([-0.1*nepochs,1.1*nepochs])
plt.yscale("log")
plt.title("Neuron Training Progress")
plt.grid()

plt.subplot(2,1,2)

plt.plot(estore,Wstore, label="W")
plt.plot(estore,bstore, label="b")
plt.xlabel("epoch")
plt.ylabel("parameter")
plt.axis([-0.1*nepochs,1.1*nepochs,0.0,3.0])
plt.grid()
plt.legend()


plt.tight_layout()
plt.show()
```
