```python
%reset -f

import numpy as np
from numpy import random
import scipy.stats
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams.update({'font.size': 16})
```

```python
### store data in arrays ###

year = np.array([1,2,3,4,5])
joke = np.array([12,18,33,27,39])
rate = np.array([5.1,4.1,3.8,3.3,3.6])
N    = len(year)
```

```python
### quick plot of the data ###

plt.scatter(joke,rate)
plt.xlabel("jokes told")
plt.ylabel("instructor rating")
plt.axis([0.0,50.0,0.0,6.0])
plt.show()
```

```python
### compute averages (need for covariance matrix terms) ###

meanx = np.sum(joke)/N
meany = np.sum(rate)/N

### vs let numpy do it ###
    
meanx_py = np.average(joke)
meany_py = np.average(rate)
```

```python
### compute covariance matrix terms ###

sigx2 = np.sum((joke-meanx)**2)/(N-1)
sigy2 = np.sum((rate-meany)**2)/(N-1)
sigx  = np.sqrt(sigx2)
sigy  = np.sqrt(sigy2)
sigxy = np.sum((joke-meanx)*(rate-meany))/(N-1)

### vs let numpy do it ###

covariance = np.cov(joke,rate)

sigx_py  = np.sqrt(covariance[0,0])
sigy_py  = np.sqrt(covariance[1,1])
sigxy_py = covariance[0,1]
```

```python
### compute Pearson correlation coefficient, p-value ### 

r = sigxy/(sigx*sigy)
p = 0.10               # read off of table in Taylor text

### vs let scipy do it ###

r_py, p_py = scipy.stats.pearsonr(joke,rate)
```

```python
### output results ###

print(" ")
print(" quantity  my calculation   numpy/scipy  ")
print("----------------------------------------")
print("   Mean x      %5.2f          %5.2f" % (meanx,meanx_py) )
print("   Mean y      %5.2f          %5.2f" % (meany,meany_py) )
print("  sigma_x      %5.2f          %5.2f" % (sigx, sigx_py)  )
print("  sigma_y      %5.2f          %5.2f" % (sigy, sigy_py)  )
print(" sigma_xy      %5.2f          %5.2f" % (sigxy, sigxy_py))
print("        r      %5.2f          %5.2f" % (r, r_py)        )
print("        p      %5.2f          %5.2f" % (p, p_py)        )
print(" ")
```

Advice to instructor: The Pearson coefficient is negative, indicating that the more jokes you tell the worse your instructor rating. The p-value of the null hypothesis (no correlation) is not that small, however, so don't feel too bad about it!
