### a) calculate Fisher information for n binomial RVs ### 

We can write down the likelihood function directly from the PDF

$$
L = \frac{n!}{k!(n-k)!} p^k (1-p)^{n-k}
$$

then the log-likelihood will be

$$
l = \ln{L} = \ln{\left(\frac{n!}{k!(n-k)!}\right)} + k \ln{p} + (n-k) \ln{(1-p)}
$$

It's derivative with respect to the bias is

$$
\frac{\partial l}{\partial p} = \frac{k}{p} - \frac{(n-k)}{(1-p)}
$$

and

$$
\left( \frac{\partial l}{\partial p} \right)^2 = \frac{k^2}{p^2} + \frac{(n-k)^2}{(1-p)^2} - 2 \frac{k}{p}\frac{(n-k)}{(1-p)}
$$

The next step is to compute the average of $\left( \partial l/\partial p \right)^2$. First, note that the RHS above is a second order polynomial in $k$. Also, anticipating that we are ultimately interested in Fisher information, and that $I_n = n I_1$, we can perform the averaging for the $n=1$, $k=0,1$ case. 

$$
\langle k \rangle~= 0 \times ({\rm probability}~k=0) + 1 \times ({\rm probability}~k = 1) = p \\
\langle k^2 \rangle~= 0^2 \times ({\rm probability}~k=0) + 1^2 \times ({\rm probability}~k = 1) = p
$$

So,

$$
I_1 = \left\langle \left(\frac{\partial l}{\partial p} \right)^2 \right\rangle~= \frac{p}{p^2} + \frac{(1 - 2p + p)}{(1-p)^2} - \frac{2p}{p(1-p)} + \frac{2p}{p(1-p)} = \frac{1}{p(1-p)}
$$

and finally

$$
I_n = n I_1 = \frac{n}{p(1-p)}
$$

### b) calculate the MLE $\hat{p}$ for the bias parameter $p$. ###

In the previous cell, we had calculated $\partial l / \partial p$. To compute MLE we just need to evaluate this expression at $\hat{p}$ and equate it to zero. 

$$
\left. \frac{\partial l}{\partial p} \right|_{p=\hat{p}} = \frac{k}{\hat{p}} - \frac{(n-k)}{(1-\hat{p})} = 0
$$

or

$$
\hat{p} = \frac{k}{n}
$$

as expected!

### c) Use the Cramer-Rao inequality to set a lower bound on the variance and standard deviation of $\hat{p}$. ###

The CR Inequality says that the variance of the estimator cannot be less than the inverse of Fisher information. In this case, that means

$$
\mathrm{Var}(\hat{p}) \geq \frac{1}{I_n} = \frac{p(1-p)}{n} 
$$

$$
\sigma_{\hat{p}} \geq \sqrt{\frac{p(1-p)}{n}} 
$$

```python
### d) Write Python code to perform $N$ trials, where in each trial a coin   ### 
###    with ground truth fairness $p_0$ is flipped $n$ times, and the number ###
###    of heads $k$ is recorded. Do this for different values of the         ###
###    ground truth $p_0$.                                                   ###

N  = 1000
n  = 10
p0 = np.arange(0.0,1.1,0.1)
```

```python
# create some storage arrays

avgp    = np.zeros(len(p0))
sigp    = np.zeros(len(p0))
sigp_CR = np.zeros(len(p0))
```

```python
# loop over ground truth biases, generating trials and recording results 

for i in np.arange(0,len(p0)):
    
    k = random.binomial(n, p0[i], size=N)
    phat = k/n
    
    avgp[i]    = np.mean(phat)
    sigp[i]    = np.std(phat,ddof=1)
    sigp_CR[i] = np.sqrt(p0[i]*(1.0-p0[i])/n)
```

```python
# plot means of trials as a function of p0

### e) for n = 10 and N = 1,000 (or more), plot mean phat versus p0             ###
###    between 0.0 and 1.0. Comment: Is there any evidence for bias in the MLE? ###

plt.figure(figsize=(6,6))
plt.subplot(1,1,1)

plt.errorbar(p0,avgp-p0,yerr=sigp/np.sqrt(N),fmt='o')
plt.xlabel('Ground Truth Fairness $p_0$')
plt.ylabel('$\\langle \hat{p} \\rangle - p_0$')
titletext = str(n)+" coin tosses"
plt.title(titletext)
plt.grid()

plt.show()
```

Comment: There is no evidence for bias in the determination of $\hat{p}$.

```python
### f) for n=10 and N=1,000 (or more), plot the standard deviation in  ###
###    phat versus p0 between 0.0 and 1.0. Superimpose points for the  ###
###    Cramer-Rao expectation. Comment: Is Cramer-Rao a good predictor ###
###    of the standard deviation in this case?                         ### 

plt.figure(figsize=(6,6))
plt.subplot(1,1,1)

plt.scatter(p0,sigp,   color='r',label="Observed")
plt.scatter(p0,sigp_CR,color='k',label="Cramer-Rao")

plt.xlabel('Ground Truth Bias $p_0$')
plt.ylabel('standard deviation of MLE bias')
plt.legend(loc="lower center")
titletext = str(n)+" coin tosses"
plt.title(titletext)
plt.grid()

plt.show()
```

Comment: Yes, the Cramer-Rao lower limit gives a good estimate of the standard deviation in this case. 

```python
### g) Now choose $n=100$ and repeat the study. Comment:      ###
###    What changes do you note in mean phat vs p0 and        ###
###    sigma_{phat} vs p0 plots? Can you explain the changes? ###  

N  = 1000
n  = 100
p0 = np.arange(0.0,1.1,0.1)

avgp    = np.zeros(len(p0))
sigp    = np.zeros(len(p0))
sigp_CR = np.zeros(len(p0))

for i in np.arange(0,len(p0)):
    
    k = random.binomial(n, p0[i], size=N)
    phat = k/n
    
    avgp[i]    = np.mean(phat)
    sigp[i]    = np.std(phat,ddof=1)
    sigp_CR[i] = np.sqrt(p0[i]*(1.0-p0[i])/n)
```

```python
### the avgp-p0 plot ###

plt.figure(figsize=(6,6))
plt.subplot(1,1,1)

plt.errorbar(p0,avgp-p0,yerr=sigp/np.sqrt(N),fmt='o')
plt.xlabel('Ground Truth Fairness $p_0$')
plt.ylabel('$\\langle \hat{p} \\rangle - p_0$')
titletext = str(n)+" coin tosses"
plt.title(titletext)
plt.grid()

plt.show()
```

```python
### the sigma(p) vs p0 plot ###

plt.figure(figsize=(6,6))
plt.subplot(1,1,1)

plt.scatter(p0,sigp,   color='r',label="Observed")
plt.scatter(p0,sigp_CR,color='k',label="Cramer-Rao")

plt.xlabel('Ground Truth Bias $p_0$')
plt.ylabel('standard deviation of MLE bias')
plt.legend(loc="lower center")
titletext = str(n)+" coin tosses"
plt.title(titletext)
plt.grid()

plt.show()
```

Comment: Both the mean plots indicate that the $\hat{p}$ determination improves with $n$. This is to be expected from the Cramer-Rao inequality

$$
\sigma_{\hat{p}} \geq \sqrt{\frac{p(1-p)}{n}} 
$$

in which the fluctuations are predicted to decrease as $1/\sqrt{n}$.
