```python
%reset -f 

import numpy as np
import scipy.optimize as SciOpt
import scipy.stats as SciStat
import scipy.special as SciSpec
from numpy import random as rnd
from numpy import linalg as LA
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rcParams.update({'font.size': 18})
```

#### Exercise 1 - Check that the "weighted sum of squares" follows a &chi;<sup>2</sup> distribution for normally distributed measurement error.

```python
N = 10
k = 10
Ntrials = 1000

mu = 2.0
sigma = 1.0

xi = np.zeros(Ntrials)

for i in np.arange(0,Ntrials,1):
    x = rnd.normal(mu, sigma, size=N)
    z = (x-mu)/sigma
    xi[i] = np.sum(z*z)   

n,b,p = plt.hist(xi,density=True,label="$\\xi$ Distribution")

xx = np.arange(0.0,30.0,0.001)
dist = SciStat.chi2(k)
yy   = dist.pdf(xx)
plt.plot(xx,yy,label="$N_{df}$ = %2d $\\chi^2$ PDF" % k)
plt.xlabel("$\\xi$")
plt.ylabel("$PDF$")
plt.axis([0.0,30.0,0.0,0.12])
plt.legend(loc="upper right")

plt.show()
```

#### Suppose that, instead of taking mu as fixed we determine mu by taking the mean of the data points. Since we have determined mu from the data, the number of degrees of freedom drops to k = N - 1.

```python
N = 10
k =  9
Ntrials = 1000

xi = np.zeros(Ntrials)

for i in np.arange(0,Ntrials,1):
    x = rnd.normal(mu, sigma, size=N)   # compute mean from the data
    mu_exp = np.mean(x)
    z = (x-mu_exp)/sigma
    xi[i] = np.sum(z*z)   

n,b,p = plt.hist(xi,density=True,label="Empirical $\\xi$")

xx = np.arange(0.0,30.0,0.001)
dist = SciStat.chi2(k)
yy   = dist.pdf(xx)
plt.plot(xx,yy,label="k = %1d PDF" % k)
plt.xlabel("$\\xi$")
plt.ylabel("$PDF$")
plt.axis([0.0,30.0,0.0,0.12])
plt.legend(loc="upper right")

plt.show()
```

#### Exercise 2: As a check of understanding how to use the scipy tools, reproduce goodness vs &chi;<sup>2</sup> plot.

```python
import scipy.special

chi2 = np.arange(0.1,100.0,0.01)

plt.figure(figsize=(8,8))

for ndf in [1,2,3,4,6,8,10,15,20,25,30,40,50]:
    goodness = scipy.special.gammaincc(0.5*ndf,0.5*chi2)
    labeltxt = str(ndf)
    plt.plot(chi2,goodness,label=labeltxt)

plt.xscale("log")
plt.yscale("log")
plt.xlabel("$\\chi^2$")
plt.ylabel("Goodness-of-Fit")
plt.axis([1.0,100.0,0.001,1.0])
plt.title("Goodness-of-Fit vs $\\chi^2$, various $N_{df}$")
plt.legend(loc="lower left")
plt.show()
```
